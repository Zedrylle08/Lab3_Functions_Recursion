```python
LAST_NAME = "Adonis"
STUDENT_ID = "TUPM-26-1828"  

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Student:", LAST_NAME)
print("Seed Digit:", SEED_DIGIT)
print("Digit Sum:", ID_SUM)
print("Surname Length:", NAME_LENGTH)
```

```python
# Task 5: Functional Encapsulation

def greet():
    print("=" * 40)
    print(f"System Initialized for: {LAST_NAME}")
    print("=" * 40)

def identity_code():
    code = SEED_DIGIT * NAME_LENGTH
    return code

# Execute functions
greet()
print("Identity Code:", identity_code())
```

```python
# Task 6: Scope and the LEGB Rule

global_value = ID_SUM

def scope_test():
    # Local scope variable shadowing global variable
    global_value = SEED_DIGIT
    print("Inside scope_test() [Local]:", global_value)

scope_test()
print("Outside scope_test() [Global]:", global_value)
```

```python
# Task 7: Parametric Processing (*args and **kwargs)

def user_summary(title, *scores, **info):
    print(f"=== {title} ===")
    total = sum(scores)
    average = total / len(scores)
    
    print("Scores:", scores)
    print("Total:", total)
    print("Average:", round(average, 2))
    
    for key, value in info.items():
        print(f"{key.capitalize()}: {value}")
        
    return average

# Function execution
avg_score = user_summary(
    f"{LAST_NAME} Academic Report",
    SEED_DIGIT * 10,
    ID_SUM % 100,
    NAME_LENGTH * 5,
    id=STUDENT_ID,
    surname=LAST_NAME
)
```

```python

def compute_area(radius):
    pi = 3.1416
    area = pi * (radius ** 2)
    return area

calculated_area = compute_area(SEED_DIGIT + 2)
print("Radius:", SEED_DIGIT + 2)
print("Computed Area:", round(calculated_area, 2))
```

```python


def safe_calculator(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        result = "UNDEFINED (Division by zero)"
    return result

print("Calculation Result (ID_SUM / SEED_DIGIT):", safe_calculator(ID_SUM, SEED_DIGIT))
print("Division by Zero Test:", safe_calculator(ID_SUM, 0))
```

```python
# Task 10: Recursive Factorial Logic

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

# Execute factorial using SEED_DIGIT
print("Factorial of Seed Digit:", factorial(SEED_DIGIT))
```

```python
# Task 11: Recursive String Reversal

def reverse_string(text):
    if len(text) <= 1:
        return text
    return text[-1] + reverse_string(text[:-1])

# Execute string reversal on surname
print("Reversed Surname:", reverse_string(LAST_NAME.upper()))
```

```python
# Task 12: Decorators

def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Executing function: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def multiply(x, y):
    return x * y

# Execute decorated function
print("Multiplication Result:", multiply(SEED_DIGIT, NAME_LENGTH))
```

```python
# Task 13: Lambda Functions & Higher-Order Functions

# Generate a list of numbers based on SEED_DIGIT
data_list = list(range(1, SEED_DIGIT + 5))

# Use lambda with map to square each number
squared_data = list(map(lambda x: x**2, data_list))

print("Original Data:", data_list)
print("Squared Data:", squared_data)
```

```python
# Task 14: Generators

def even_squares(limit):
    for i in range(limit):
        if i % 2 == 0:
            yield i**2

# Execute generator using NAME_LENGTH + SEED_DIGIT as limit
limit_val = NAME_LENGTH + SEED_DIGIT
print(f"Even squares up to limit {limit_val}:")
for value in even_squares(limit_val):
    print(value)
```
